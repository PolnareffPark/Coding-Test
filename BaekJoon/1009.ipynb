문제
재용이는 최신 컴퓨터 10대를 가지고 있다. 어느 날 재용이는 많은 데이터를 처리해야 될 일이 생겨서 각 컴퓨터에 1번부터 10번까지의 번호를 부여하고, 10대의 컴퓨터가 다음과 같은 방법으로 데이터들을 처리하기로 하였다.

1번 데이터는 1번 컴퓨터, 2번 데이터는 2번 컴퓨터, 3번 데이터는 3번 컴퓨터, ... ,

10번 데이터는 10번 컴퓨터, 11번 데이터는 1번 컴퓨터, 12번 데이터는 2번 컴퓨터, ...

총 데이터의 개수는 항상 a^b개의 형태로 주어진다. 재용이는 문득 마지막 데이터가 처리될 컴퓨터의 번호가 궁금해졌다. 이를 수행해주는 프로그램을 작성하라.

입력
입력의 첫 줄에는 테스트 케이스의 개수 T가 주어진다. 그 다음 줄부터 각각의 테스트 케이스에 대해 정수 a와 b가 주어진다. (1 ≤ a < 100, 1 ≤ b < 1,000,000)

출력
각 테스트 케이스에 대해 마지막 데이터가 처리되는 컴퓨터의 번호를 출력한다.

```python
# 시간 초과가 발생합니다.

num_test_case = int(input())

for _ in range(num_test_case):
    A, B = map(int, input().split())
    print(A**B%10)
```

```python
# 제곱을 하지 않고도 맨 마지막 숫자를 찾아내는 것이 과제의 핵심으로 보여집니다
# 또 시간 초과가 발생했습니다.

num_test_case = int(input())

for _ in range(num_test_case):
    A, B = map(int, input().split())
    rest = A
    if B >=1:
        for i in range(B-1):
            rest = (rest * A)%10
        print(rest)
    else:
        print(rest)
```

```python
# 결국엔 모든 경우의 수를 구해서 풀었습니다

num_2 = [2, 4, 8, 6]
num_3 = [3, 9, 7, 1]
num_4 = [4, 6]
num_7 = [7, 9, 3, 1]
num_8 = [8, 4, 2, 6]
num_9 = [9, 1]

num_test_case = int(input())

for _ in range(num_test_case):
    A, B = map(int, input().split())
    
    if A % 10 == 0:
        print(10)
    else:
        A = A % 10
        
        if 1 == A or 5 == A or 6 == A:
            print(A)
        elif A == 2:
            print(num_2[B%len(num_2)-1])
        elif A == 3:
            print(num_3[B%len(num_3)-1])
        elif A == 4:
            print(num_4[B%len(num_4)-1])
        elif A == 7:
            print(num_7[B%len(num_7)-1])
        elif A == 8:
            print(num_8[B%len(num_8)-1])
        elif A == 9:
            print(num_9[B%len(num_9)-1])
```

```python
# 더 쉬운 방법이 있었습니다

t = int(input())
for _ in range(t):
    a, b = map(int, input().split())
    last = pow(a, b, 10)     # a^b의 일의 자리
    print(10 if last == 0 else last)
```

map(함수, 반복가능한 데이터) 는

"반복 가능한 데이터" 를 "함수"를 적용해서 출력합니다.

pow 함수는 다음과 같습니다.

```python
pow(3, 4) # 3**4 반환
pow(3, 4, 10) # 3**4%10 을 반환
```
