문제
- 알파벳 대소문자로 된 단어가 주어지면, 이 단어에서 가장 많이 사용된 알파벳이 무엇인지 알아내는 프로그램을 작성하시오. 단, 대문자와 소문자를 구분하지 않는다.

입력
- 첫째 줄에 알파벳 대소문자로 이루어진 단어가 주어진다. 주어지는 단어의 길이는 1,000,000을 넘지 않는다.

출력
- 첫째 줄에 이 단어에서 가장 많이 사용된 알파벳을 대문자로 출력한다. 단, 가장 많이 사용된 알파벳이 여러 개 존재하는 경우에는 ?를 출력한다.

````python
# 최종 제출한 풀이
# 이 문제는 시간 안에 풀어야 하는 게 어렵습니다. 그러니 이중반복은 할 수 없습니다.

data = input().upper()

dics = {}

for spell in data:
    if spell in dics:
        continue
    dics[spell] = data.count(spell)

values = []
rev_dics = {}

for keys in dics:
    values.append(dics[keys])
    rev_dics[dics[keys]] = keys

if values.count(max(values)) != 1:
    print('?')
else:
    print(rev_dics[max(values)])
````

## `collections`

### `Counter` — 빈도수(개수) 세기

* **용도:** iterable 내 원소의 출현 횟수를 사전 형태로 집계
* **주요 기능:** `most_common`, 딕셔너리처럼 인덱싱 가능

```python
from collections import Counter

c = Counter("MISSISSIPPI")
print(c)                     # Counter({'I':4, 'S':4, 'P':2, 'M':1})
print(c.most_common(1))      # [('I', 4)]  (최빈 1개)
print(c['S'])                # 4
```

---

### `deque` — 양끝 O(1) 삽입/삭제 큐

* **용도:** 큐/덱 구현, 슬라이딩 윈도우, BFS
* **주요 기능:** `append`, `appendleft`, `pop`, `popleft`, `rotate`

```python
from collections import deque

dq = deque([1, 2, 3])
dq.append(4)      # 오른쪽 끝에 추가
dq.appendleft(0)  # 왼쪽 끝에 추가
print(dq)         # deque([0, 1, 2, 3, 4])
dq.popleft()
print(dq)         # deque([1, 2, 3, 4])
dq.rotate(1)
print(dq)         # deque([4, 1, 2, 3])
```

---

### `defaultdict` — 기본값이 있는 dict

* **용도:** 키가 없을 때 자동으로 기본값 생성(예: 리스트 누적)
* **주요 기능:** 생성 시 `default_factory` 지정

```python
from collections import defaultdict

dd = defaultdict(list)
dd['a'].append(1)
dd['a'].append(2)
dd['b'].append(3)
print(dd)         # defaultdict(<class 'list'>, {'a':[1,2], 'b':[3]})
print(dd['c'])    # []  (접근 시 자동 생성)
```

## `heapq`

### 최소 힙(우선순위 큐)

* **용도:** 최솟값을 빠르게 꺼내기(로그 시간), K개 유지, 정렬 대체
* **주의:** 파이썬 힙은 **최소 힙**. 최대 힙은 음수화하거나 `(key, item)` 트릭 사용.

```python
import heapq

h = []
heapq.heappush(h, 3)
heapq.heappush(h, 1)
heapq.heappush(h, 2)
print(h)                   # [1, 3, 2]  (힙 내부 표현)
print(heapq.heappop(h))    # 1
print(heapq.heappop(h))    # 2
print(heapq.heappop(h))    # 3
```

## `bisect`

### 이분 탐색 인덱스 (`bisect_left`, `bisect_right`)

* **용도:** 정렬된 배열에 원소 삽입 위치, 구간 개수 세기
* `bisect_left(a, x)`: x를 넣을 **왼쪽** 위치
* `bisect_right(a, x)`: x를 넣을 **오른쪽** 위치

```python
import bisect

a = [1, 2, 2, 2, 5]
print(bisect.bisect_left(a, 2))   # 1  (2의 시작 위치)
print(bisect.bisect_right(a, 2))  # 4  (2의 끝 다음 위치)
# 배열에 x를 정렬상태 유지하며 삽입
bisect.insort(a, 3)
print(a)                          # [1, 2, 2, 2, 3, 5]
```

## `functools`

### `lru_cache` — 메모이제(재귀/DP 최적화)

* **용도:** 동일 인자 호출 결과 캐싱 → 중복 계산 제거

```python
from functools import lru_cache

@lru_cache(maxsize=None)
def fib(n):
    if n < 2: return n
    return fib(n-1) + fib(n-2)

print(fib(10))     # 55
print(fib.cache_info())  # CacheInfo(hits=..., misses=..., maxsize=None, currsize=...)
```

### `cmp_to_key` — 정렬 비교함수 어댑터

* **용도:** `sorted(..., key=...)`만 지원하는 파이썬에서 C 스타일 비교함수를 사용하고 싶을 때

```python
from functools import cmp_to_key

def cmp(a, b):
    # 내림차순: a>b면 앞에
    return (b - a)

arr = [3, 1, 4, 2]
print(sorted(arr, key=cmp_to_key(cmp)))
# [4, 3, 2, 1]
```

> 팁: 대부분은 `key` 기반 정렬이 더 간단/빠릅니다. `cmp_to_key`는 복합 비교(여러 필드, 맞춤 규칙)가 필요할 때만 사용하세요.
