문제
- 8진수가 주어졌을 때, 2진수로 변환하는 프로그램을 작성하시오.

입력
- 첫째 줄에 8진수가 주어진다. 주어지는 수의 길이는 333,334을 넘지 않는다.

출력
- 첫째 줄에 주어진 수를 2진수로 변환하여 출력한다. 수가 0인 경우를 제외하고는 반드시 1로 시작해야 한다.

```python
data = input()

if data=='0':
    print(data)
    
else:
    ans = []

    for i in data:
        i = int(i)
        if i-4<0:
            ans.append('0')
        else:
            i -= 4
            ans.append('1')
        
        if i-2<0:
            ans.append('0')
        else:
            i -= 2
            ans.append('1')
        
        if i-1<0:
            ans.append('0')
        else:
            ans.append('1')

    ans = ''.join(ans)
    while ans[0]=='0':
        ans = ans[1:]
    print(ans)
```

```python
# 훠어어어어얼씬 더 짧은 결과가 있습니다.

# input 으로 받은 8진수 문자열을 -> 10진수로 바꾸고 -> f"{정수:b}" 또는 format(정수, 'b') 로 이진화
print(f'{int(input(),8):b}')

# 입력 -> 10진 변환 -> bin(정수) -> 2진 변환 -> [2:]으로 앞의 '2b'을 떼고 사용
print(bin(int(input(),8))[2:])
```
