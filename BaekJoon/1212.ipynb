문제
- 8진수가 주어졌을 때, 2진수로 변환하는 프로그램을 작성하시오.

입력
- 첫째 줄에 8진수가 주어진다. 주어지는 수의 길이는 333,334을 넘지 않는다.

출력
- 첫째 줄에 주어진 수를 2진수로 변환하여 출력한다. 수가 0인 경우를 제외하고는 반드시 1로 시작해야 한다.

```python
data = input()

if data=='0':
    print(data)
    
else:
    ans = []

    for i in data:
        i = int(i)
        if i-4<0:
            ans.append('0')
        else:
            i -= 4
            ans.append('1')
        
        if i-2<0:
            ans.append('0')
        else:
            i -= 2
            ans.append('1')
        
        if i-1<0:
            ans.append('0')
        else:
            ans.append('1')

    ans = ''.join(ans)
    while ans[0]=='0':
        ans = ans[1:]
    print(ans)
```

```python
# 훠어어어어얼씬 더 짧은 결과가 있습니다.

# input 으로 받은 8진수 문자열을 -> 10진수로 바꾸고 -> f"{정수:b}" 또는 format(정수, 'b') 로 이진화
print(f'{int(input(),8):b}')

# 입력 -> 10진 변환 -> bin(정수) -> 2진 변환 -> [2:]으로 앞의 '2b'을 떼고 사용
print(bin(int(input(),8))[2:])
```

```python
a = int('4242', 8)
print(f'a: {a}')
print(f"a(2) : {a:b}")

a = int('4242', 8)
print(f'a: {a}')
print(f"a(2) : {bin(a)}")
```

```python
# 포맷 지정자 자주 쓰는 것 예시

a = 42
print(f"{a:b}")   # 2진수 -> '101010'
print(f"{a:o}")   # 8진수 -> '52'
print(f"{a:d}")   # 10진수 -> '42'
print(f"{a:x}")   # 16진수(소문자) -> '2a'
print(f"{a:X}")   # 16진수(대문자) -> '2A'

print(f"{a:#b}")  # '0b101010'
print(f"{a:#x}")  # '0x2a'

n = 7
print(f"{n:>5}")      # 우측 정렬, 폭 5: '    7'
print(f"{n:<5}")      # 좌측 정렬: '7    '
print(f"{n:^5}")      # 가운데 정렬: '  7  '
print(f"{n:0>5}")     # 빈칸을 '0'으로 채우기: '00007'

x, y = 123, -123
print(f"{x:+d}")   # '+123'
print(f"{y:+d}")   # '-123'
print(f"{x:05d}")  # '00123' (폭 5, 0패딩)

big = 1234567890
print(f"{big:,}")   # '1,234,567,890'
print(f"{big:_}")   # '1_234_567_890' (언더스코어)

s = "hi"
print(f"{s:>5}")    # '   hi'  (우측정렬)
print(f"{s:*^6}")   # '**hi**' (가운데 정렬, 채움문자 '*', 폭 6)
print(f"{s:.1}")    # 'h'      (최대 길이 1)

obj = "line\nbreak"
print(f"{obj!s}")  # 실제 출력: 줄바꿈 포함
print(f"{obj!r}")  # 리터럴 표현: 'line\nbreak'

a = 42
print(format(a, "b"))     # '101010'
print("a(2) : {}".format(format(a, "b")))

n = 255
print(f"{n:#010x}")   # 접두어 포함 16진수, 폭 10, 0패딩: '0x000000ff'
print(f"{n:=+8d}")    # 부호를 맨 앞에 고정(=), 폭 8: '    +255'
```
