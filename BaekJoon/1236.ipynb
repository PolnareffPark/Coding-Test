문제
- 영식이는 직사각형 모양의 성을 가지고 있다. 성의 1층은 몇 명의 경비원에 의해서 보호되고 있다. 영식이는 모든 행과 모든 열에 한 명 이상의 경비원이 있으면 좋겠다고 생각했다.

- 성의 크기와 경비원이 어디있는지 주어졌을 때, 몇 명의 경비원을 최소로 추가해야 영식이를 만족시키는지 구하는 프로그램을 작성하시오.

입력
- 첫째 줄에 성의 세로 크기 N과 가로 크기 M이 주어진다. N과 M은 50보다 작거나 같은 자연수이다. 둘째 줄부터 N개의 줄에는 성의 상태가 주어진다. 성의 상태는 .은 빈칸, X는 경비원이 있는 칸이다.

출력
- 첫째 줄에 추가해야 하는 경비원의 최솟값을 출력한다.

```python
row, col = map(int, input().split())

rows = []

for _ in range(row):
    row_data = input()
    rows.append(row_data)

cols = list(zip(*rows))

row_count = col_count = 0

for row_data in rows:
    if 'X' in row_data:
        row_count += 1

for col_data in cols:
    if 'X' in col_data:
        col_count += 1

print(max(row-row_count, col-col_count))
```

```python
# 내 풀이 업데이트 : 더 직관적이고 한 번에 끝내는 방법

row, col = map(int, input().split())

cols_has = [False] * col   # 각 열에 'X'가 한 번이라도 나왔는지
rows_needed = 0

for _ in range(row):
    line = input().strip()
    has_X = False
    for j, ch in enumerate(line):
        if ch == 'X':
            has_X = True
            cols_has[j] = True
    if not has_X:          # 이 행에 X가 한 번도 없으면, 경비원 필요
        rows_needed += 1

cols_needed = cols_has.count(False)  # X가 한 번도 없던 열의 개수
print(max(rows_needed, cols_needed))
```
