문제
- N개의 정수가 주어지면, 이 정수들의 합 S의 부호를 구하는 프로그램을 작성하시오.

입력
- 총 3개의 테스트 셋이 주어진다. 각 테스트 셋의 첫째 줄에는 N(1 ≤ N ≤ 100,000)이 주어지고, 둘째 줄부터 N개의 줄에 걸쳐 각 정수가 주어진다. 주어지는 정수의 절댓값은 9223372036854775807보다 작거나 같다.

출력
- 총 3개의 줄에 걸쳐 각 테스트 셋에 대해 N개의 정수들의 합 S의 부호를 출력한다. S=0이면 "0"을, S>0이면 "+"를, S<0이면 "-"를 출력하면 된다.

```python
def sol():
    N = int(input())
    data = []
    for _ in range(N):
        info = int(input())
        data.append(info)
    sums = sum(data)
    if sums>0:
        print('+')
    elif sums<0:
        print('-')
    else:
        print(0)

sol()
sol()
sol()
```

```python
# 위 코드를 더 간단히 정리

def main():
    for _ in range(3): # 3번 들어가는 게 정해져 있으므로 설정
        N = int(input()) # N 값 입력
        s = 0
        for _ in range(N): # N만큼 반복 입력
            s += int(input()) # s 에 누적합
        print(0 if s == 0 else ('+' if s > 0 else '-'))

main()
```

```python
# 가장 깔끔한 스트리밍 풀이

def main():
    it = iter(map(int, open(0)))  # 표준입력의 숫자들을 순서대로 뽑는 이터레이터
    for _ in range(3):
        N = next(it)
        s = 0
        for _ in range(N):
            s += next(it)
        print(0 if s == 0 else ('+' if s > 0 else '-'))

main()
```

```python
# 슬라이싱 풀이

lis = [*map(int, open(0))]
for _ in range(3):
    N = lis[0]
    answer = sum(lis[1:N+1])
    lis = lis[N+1:]
    print(0 if not answer else '-' if answer < 0 else '+')
```
