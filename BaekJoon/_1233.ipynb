```python
# 위 코드를 살려서 한번 만들어봅니다.

S1, S2, S3 = map(int, input().split())

# 1) 모든 합을 모은 리스트, 
# 앞에서처럼 S1 S2 S3 값들을 모아서 리스트로 만들 필요가 없음
data = []
for a in range(1, S1+1):
    for b in range(1, S2+1):
        for c in range(1, S3+1):
            data.append(a + b + c)

# 2) 합 → 빈도 딕셔너리
freq = {}
for s in data: # s는 실제 합 값
    freq[s] = freq.get(s, 0) + 1

# 3) 최빈도 찾기
max_cnt = max(freq.values())

# 4) 최빈도의 합들 중 '가장 작은 합' 선택
# freq.items() 로 key, val 값을 받아서, val 값이 최빈도와 같은 key 를 candidate 에 모음
candidates = [s for s, cnt in freq.items() if cnt == max_cnt]

# 모아진 candidate 중 최소값을 출력
print(min(candidates))
```

```python
# value 가 key, key 가 value 인 역딕셔너리 만들기.
# freq: {'a' = 1, 'b' = 2, 'c' = 3}
rev = {}
for s, cnt in freq.items():
    rev.setdefault(cnt, []).append(s)
# rev = {1 = 'a', 2 = 'b', 3 = 'c'}
# setdefault 는 cnt 가 rev 에 있으면 value를, 없으면 [] 를 반환합니다.

# value 가 최대인 key 를 꺼내는 간단 패턴들
key_with_max = max(freq, key=freq.get) # 가장 큰 value 를 갖는 key 하나(동률 무시)
max_cnt = max(freq.values()) # 가장 큰 value(최빈도) 자체
winners = [k for k, v in freq.items() if v == max_cnt] # 최빈도를 가진 key 전부
answer = min(winners) # 최빈도 key 들중 가장 작은 key
```

이런 전형적인 빈도 카운트 문제에 대한 접근 방법 for Chat GPT

- 리스트로 빈도 배열을 만든다. 딕셔너리는 필요없다!
- 삼중 루프로 합을 만들며 지나간다.
- max 값을 구하고, 가장 작은 인덱스에서 max 가 나오는 위치를 찾는다.

```python
S1, S2, S3 = map(int, input().split())

# 합의 최댓값 + 1 크기 배열 (인덱스로 합을 바로 사용)
max_sum = S1 + S2 + S3
cnt = [0] * (max_sum + 1)

for a in range(1, S1 + 1):
    for b in range(1, S2 + 1):
        for c in range(1, S3 + 1):
            cnt[a + b + c] += 1

max_cnt = max(cnt)
# 가장 작은 합(인덱스)부터 찾기
for s in range(3, max_sum + 1):
    if cnt[s] == max_cnt:
        print(s)
        break
```

"인덱스가 너무 크거나 케이스가 너무 많아서 1초 안에 전부 순회 못하는" 

상황에서 자주 쓰는 속도 올리는 4가지 접근 패턴

```python
# 1. 수학적 축약 : 이중합/삼중합을 전개해 "합의 곱"같은 형태로 바꿔 한 번에 계산.
# 예제: 문자열 A, B 의 자리수 쌍곱 합

A, B = "3 5".split()
ans = sum(map(int, A)) * sum(map(int, B))
print(ans)
```

```python
# 2. 좌표 압축 : 값 범위가 크지만 서로 다른 값은 적을 때.
# 값이 최대 10^9인 수열의 등장 빈도 세기

arr = list(map(int, "5 100 5 7 100 5".split()))
# arr = [5, 100, 5, 7, 100, 5]

unique = sorted(set(arr))
# unique = [5, 7, 100]

idx = {v:i for i,v in enumerate(unique)}
# idx = {5:0, 7:1, 100:2}

cnt = [0] * len(unique)
# cnt = [0 0 0]

for x in arr: #  [5, 100, 5, 7, 100, 5] 입력
    cnt[idx[x]] += 1 # cnt[idx[5]] == cnt[0]

# unique = [5, 7, 100]
# cnt = [3, 1, 2] -> arr 에 5가 3번, 7이 1번, 100이 2번 등장했다는 의미
print(cnt)
```

```python
# 투 포인터 / 슬라이딩 윈도우 (단조/연속 구간 문제)
# 정렬된 배열이나 윈도우 내 조건을 유지하며 포인터를 한 방향으로만 움직여 O(n)에 해결
# 목표 : nums 에서 원소 합이 K 이하인 가장 긴 연속 구간 길이

nums = list(map(int, "2 1 5 1 3 2".split()))
# nums = [2, 1, 5, 1, 3, 2]

K = int("7")
# K = 7

best = s = L = 0
# best: 지금까지 본 최장 길이
# s = 현재 윈도우 합
# L, R : 현재 윈도우의 좌우 인덱스

# [(0, 2), (1, 1), (2, 5), (3, 1), (4, 3), (5, 2)]
for R, x in enumerate(nums):
    s += x
    while s > K:
        s -= nums[L]
        L += 1
    best = max(best, R-L+1)
"""
| R | x | s(더한 후) | while 줄이기             |줄인뒤 s | L | 윈도우 | 길이 | best |
| - | - | ------- | -------------------------- | ------ | - | ------- | -- | ---- |
| 0 | 2 | 2       | (s≤7, 안 줄임)             | 2      | 0 | [2]     | 1  | 1    |
| 1 | 1 | 3       | (s≤7)                      | 3      | 0 | [2,1]   | 2  | 2    |
| 2 | 5 | 8       | 8>7 → s-=2,L=1             | 6      | 1 | [1,5]   | 2  | 2    |
| 3 | 1 | 7       | (s≤7)                      | 7      | 1 | [1,5,1] | 3  | 3    |
| 4 | 3 | 10      | 10>7→s-=1,L=2=9>7→s-=5,L=3 | 4      | 3 | [1,3]   | 2  | 3    |
| 5 | 2 | 6       | (s≤7)                      | 6      | 3 | [1,3,2] | 3  | 3    |
"""

print(best)
```

```python
# 프리픽스 합 + 해시
# 예제: 부분합이 K인 구간 수 세기(음수 포함 O(n))

nums = list(map(int, "1 -1 2 3".split()))
# nums = [1, -1, 2, 3]

K = int("3")
# K = 3

"""
일반 dict는 없는 키를 꺼내면 에러가 발생
d = {}
d['a'] += 1 # KeyError : 'a' 키가 아직 없음

그래서 보통은 카운팅 코드를 이렇게 우회함
d = {}
d['a'] = d.get(a, 0) + 1 # get 으로 기본값 0을 지정

defaultdict 는 "숫자 카운터용 딕셔너리" 를 자동으로 만들어 줌!
cnt = defaultdict(int)
cnt['a'] += 1 # KeyError 안 발생
cnt['b'] += 2
print(cnt) # defaultdict(<class: 'int'>, {'a' : 1, 'b' : 2})
"""
from collections import defaultdict
cnt = defaultdict(int)

cnt[0] = 1 # {0 : 1}, 시작점 처리

ans = s = 0

for x in nums: # nums : [1, -1, 2, 3]
    s += x # s = 0 + 1 = 1
           # s = 1 - 1 = 0
           # s = 0 + 2 = 2
           # s = 2 + 3 = 5
    ans += cnt[s - K] # ans += cnt[1 - 3] = cnt[-2] = 0
                      # ans += cnt[0 - 3] = cnt[-3] = 0
                      # ans += cnt[2 - 3] = cnt[-1] = 0
                      # ans += cnt[5 - 3] = cnt[2] = 1
    cnt[s] += 1 # cnt[1] = cnt[1] + 1 -> cnt = {0:1, 1:1}
                # cnt[0] = cnt[0] + 1 -> cnt = {0:2, 1:1}
                # cnt[2] = cnt[2] + 1 -> cnt = {0:2, 1:1, 2:1}
                # cnt[5] = cnt[5] + 1 -> cnt = {0:2, 1:1, 2:1, 5:1}
print(ans)
```
