Write a function to find the longest common prefix string amongst an array of strings.

If there is no common prefix, return an empty string "".

 

Example 1:

Input: strs = ["flower","flow","flight"]
Output: "fl"
Example 2:

Input: strs = ["dog","racecar","car"]
Output: ""
Explanation: There is no common prefix among the input strings.
 

Constraints:

1 <= strs.length <= 200
0 <= strs[i].length <= 200
strs[i] consists of only lowercase English letters if it is non-empty.

```python
# 틀린 솔루션

class Solution:
    def longestCommonPrefix(self, strs: List[str]) -> str:
        first_word = strs[0]
        for word in strs: # 이렇게 하면 첫 번째 단어를 또 받아서 쓸모 없는 반복이 생깁니다.
            while first_word not in word: # while 문이 잘못되었습니다.
                first_word = first_word[:-1]
                if not first_word:
                    break
        return first_word
```

```python
# 제출 답안

class Solution:
    def longestCommonPrefix(self, strs: List[str]) -> str:
        spells = list(zip(*strs))
        answer = ''
        for _, spell in enumerate(spells): # 바로 for i in zip(*strs) 가 더 간결합니다.
            if len(set(spell)) == 1:
                answer += spell[0]
            else:
                break
        return answer
```

```python
# 2번째 해답 - 수직 스캔

class SolutionVerticalZip:
    def longestCommonPrefix(self, strs: list[str]) -> str:
        if not strs:
            return ""
        res = []
        
        for column in zip(*strs):    # 각 위치의 문자 튜플들
            
            # 집합 자료형 set 은 "중복을 허용하지 않음" + "순서가 없음" 이므로
            # set 으로 같은 정보들 ['f', 'f', 'f'] 를 묶으면 1개만 남게 됩니다.
            if len(set(column)) == 1:
                res.append(column[0])
            else:
                break
            
        return "".join(res)
```
