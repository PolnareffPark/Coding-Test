Given a string s containing just the characters '(', ')', '{', '}', '[' and ']', determine if the input string is valid.

An input string is valid if:

Open brackets must be closed by the same type of brackets.
Open brackets must be closed in the correct order.
Every close bracket has a corresponding open bracket of the same type.
 

Example 1:

Input: s = "()"

Output: true

Example 2:

Input: s = "()[]{}"

Output: true

Example 3:

Input: s = "(]"

Output: false

Example 4:

Input: s = "([])"

Output: true

Example 5:

Input: s = "([)]"

Output: false

 

Constraints:

1 <= s.length <= 10^4
s consists of parentheses only '()[]{}'.

```python
class Solution:
    def isValid(self, s: str) -> bool:
        # 길이가 홀수면 바로 불가능
        # 괄호는 항상 2개가 쌍이기 때문에 항상 짝수입니다.
        if len(s) % 2 == 1:
            return False

        # 여는 괄호 -> 기대하는 닫는 괄호
        expect = {'(' : ')', '[' : ']', '{' : '}'}
        stack = []

        for ch in s:
            if ch in expect:                 # 여는 괄호 (, [, { 가 나왔을 때
                stack.append(expect[ch])     # 기대하는 '닫는 괄호'를 푸시
                print(f"stack: {stack}")
            else:                            # 닫는 괄호
                print(f"ch: {ch}")  # 다음에 도착할 닫는 괄호를 기대합니다.
                # 닫는 괄호를 내야 하는데 stack이 비어 있거나,
                # 예상하는 닫는 괄호가 stack에서 꺼낸 것과 다르다면 False 를 반환합니다.
                if not stack or stack.pop() != ch:
                    return False

        return not stack # 모든 괄호를 열었다 닫았으므로 stack은 비어 있어야 합니다.
```

```python
class Solution:
    def isValid(self, s: str) -> bool:
        # 홀수일 때는 항상 False 이므로 빠른 탈출
        if len(s) % 2 == 1:
            return False

        # 닫는 괄호 -> 기대하는 여는 괄호
        need_open = {')': '(', ']': '[', '}': '{'} # 닫는 괄호를 딕셔너리로 저장합니다.
        opens = set(need_open.values()) # 여는 괄호들만 뽑아서 저장합니다. ('(', '[', '{')
        stack = [] # 스택할 빈 리스트

        for ch in s:
            if ch in opens: # 받은 문자가 여는 괄호일 때
                stack.append(ch) # 여는 괄호를 스택에 쌓습니다.
            
            else: # 받은 문자가 닫는 괄호일 때
                # 스택이 비었거나 (not stack)
                # 스택에 가장 나중에 넣은 괄호와 지금 문자열에서의 닫는 괄호가 다를 때
                # False 를 반환합니다
                if not stack or stack[-1] != need_open[ch]: 
                    return False
                
                # 넣었던 괄호를 빼냅니다.
                stack.pop()

        return not stack
```

```python
# 가장 짧은 모범 답안

class Solution:
    def isValid(self, s: str) -> bool:
        stack = []
        close_map = {
            "}": "{",
            ")": "(",
            "]": "["
        }

        for c in s:
            if c in close_map: # 만약 닫는 괄호를 받았다면
                # not stack -> stack 이 차 있으면 False, 다음 조건을 봅니다
                # "스택에서 뽑은 값"이 "받은 닫는 괄호와 매칭되는 여는 괄호"와 다르다면
                # False 를 반환합니다.
                if not stack or stack.pop() != close_map[c]:
                    return False
            else: # 여는 괄호 추가
                stack.append(c)
        
        return not stack
```
