# max()

연산 ⊕는 두 정수에 대한 연산으로 두 정수를 붙여서 쓴 값을 반환합니다. 예를 들면 다음과 같습니다.

12 ⊕ 3 = 123
3 ⊕ 12 = 312
양의 정수 a와 b가 주어졌을 때, a ⊕ b와 b ⊕ a 중 더 큰 값을 return 하는 solution 함수를 완성해 주세요.

단, a ⊕ b와 b ⊕ a가 같다면 a ⊕ b를 return 합니다.

```python
# 내 풀이

def solution(a, b):
    a_b = int(str(a) + str(b))
    b_a = int(str(b) + str(a))
    if a_b >= b_a:
        return a_b
    else:
        return b_a
```

```python
# max 를 이용하면 더 짧게 풀 수 있습니다.

def solution(a, b):
    return max(int(f"{a}{b}"), int(f"{b}{a}"))
```

# 조건 문자열

문자열에 따라 다음과 같이 두 수의 크기를 비교하려고 합니다.

두 수가 n과 m이라면

">", "=" : n >= m

"<", "=" : n <= m

">", "!" : n > m

"<", "!" : n < m

두 문자열 ineq와 eq가 주어집니다. 

ineq는 "<"와 ">"중 하나고, eq는 "="와 "!"중 하나입니다. 

그리고 두 정수 n과 m이 주어질 때, n과 m이 ineq와 eq의 조건에 맞으면 1을 아니면 0을 return하도록 solution 함수를 완성해주세요.

```python
# 이걸로 풀었습니다. eval 을 써서 푸는 방법이 있긴 하던데 보안 이슈가 크다고 합니다.

def solution(ineq, eq, n, m):
    if eq == '=':
        if ineq == '>':
            answer = n >= m
        else:
            answer = n <= m
    else:
        if ineq == '>':
            answer = n > m
        else:
            answer = n < m
    return int(answer)
```

# 코드 처리하기

문자열 code가 주어집니다.
- code를 앞에서부터 읽으면서 만약 문자가 "1"이면 mode를 바꿉니다. mode에 따라 code를 읽어가면서 문자열 ret을 만들어냅니다.

- mode는 0과 1이 있으며, idx를 0 부터 code의 길이 - 1 까지 1씩 키워나가면서 code[idx]의 값에 따라 다음과 같이 행동합니다.

mode가 0일 때
- code[idx]가 "1"이 아니면 idx가 짝수일 때만 ret의 맨 뒤에 code[idx]를 추가합니다.
- code[idx]가 "1"이면 mode를 0에서 1로 바꿉니다.

mode가 1일 때
- code[idx]가 "1"이 아니면 idx가 홀수일 때만 ret의 맨 뒤에 code[idx]를 추가합니다.
- code[idx]가 "1"이면 mode를 1에서 0으로 바꿉니다.
- 문자열 code를 통해 만들어진 문자열 ret를 return 하는 solution 함수를 완성해 주세요.

단, 시작할 때 mode는 0이며, return 하려는 ret가 만약 빈 문자열이라면 대신 "EMPTY"를 return 합니다.

```python
def solution(code):
    mode = 0
    ret = ""
    
    for i, j in enumerate(code):
        if not mode:
            if j == '1':
                mode = 1
                continue
            elif i%2 == 0:
                ret += code[i]
        else:
            if j == '1':
                mode = 0
                continue
            elif i%2 == 1:
                ret += code[i]
    
    return ret if ret else "EMPTY"
```

```python
a = solution("abc1abc1abc")
print(a)

b = solution("")
print(b)
```

# 등차수열의 특정한 항만 더하기

- 두 정수 a, d와 길이가 n인 boolean 배열 included가 주어집니다. 첫째항이 a, 공차가 d인 등차수열에서 included[i]가 i + 1항을 의미할 때, 이 등차수열의 1항부터 n항까지 included가 true인 항들만 더한 값을 return 하는 solution 함수를 작성해 주세요.

```python
# 내가 낸 풀이
def solution(a, d, included):
    AP = []
    answer = 0
    for count in range(len(included)):
        temp = a + d * count
        AP.append(temp)
        
    AP_zip = list(zip(AP, included))
    
    for i, j in enumerate(AP_zip):
        if AP_zip[i][1]:
            answer += AP_zip[i][0]
    
    return answer
```

# 주사위 게임 2

- 1부터 6까지 숫자가 적힌 주사위가 세 개 있습니다. 세 주사위를 굴렸을 때 나온 숫자를 각각 a, b, c라고 했을 때 얻는 점수는 다음과 같습니다.

- 세 숫자가 모두 다르다면 a + b + c 점을 얻습니다.
- 세 숫자 중 어느 두 숫자는 같고 나머지 다른 숫자는 다르다면 (a + b + c) × (a^2 + b^2 + c^2 )점을 얻습니다.
- 세 숫자가 모두 같다면 (a + b + c) × (a^2 + b^2 + c^2 ) × (a^3 + b^3 + c^3 )점을 얻습니다.
- 세 정수 a, b, c가 매개변수로 주어질 때, 얻는 점수를 return 하는 solution 함수를 작성해 주세요.

```python
# 제출 풀이입니다.

def solution(a, b, c):
    answer_list = list(set([a,b,c]))
    sol = 0
    if len(answer_list) == 3:
        sol = a + b + c
    elif len(answer_list) == 1:
        sol = 3*a * 3*a**2 * 3*a**3
    else:
        if not a-b:
            sol = (2*a + c) * (2*a**2 + c**2)
        elif not a-c:
            sol = (2*a + b) * (2*a**2 + b**2)
        else:
            sol = (2*b + a) * (2*b**2 + a**2)
    return sol
```

# 원소들의 곱과 합

- 정수가 담긴 리스트 num_list가 주어질 때, 모든 원소들의 곱이 모든 원소들의 합의 제곱보다 작으면 1을 크면 0을 return하도록 solution 함수를 완성해주세요.

```python
def solution(num_list):
    square = 0
    mul = 1
    
    for num in num_list:
        square += num
    square = pow(square, 2)
    
    for num in num_list:
        mul = mul * num
    
    return int(bool(mul<square))
```

# 수 조작하기

정수 n과 문자열 control이 주어집니다. control은 "w", "a", "s", "d"의 4개의 문자로 이루어져 있으며, control의 앞에서부터 순서대로 문자에 따라 n의 값을 바꿉니다.

- "w" : n이 1 커집니다.
- "s" : n이 1 작아집니다.
- "d" : n이 10 커집니다.
- "a" : n이 10 작아집니다.

위 규칙에 따라 n을 바꿨을 때 가장 마지막에 나오는 n의 값을 return 하는 solution 함수를 완성해 주세요

```python
# 내가 푼 코드

def solution(n, control):
    for word in control:
        if word == 'w':
            n += 1
        elif word == 's':
            n -= 1
        elif word == 'd':
            n += 10
        else :
            n -= 10
    return n
```

```python
# 더 파이써닉하고 유용한 코드가 있었습니다.
# 딕셔너리를 통해 저장하는 방법이 있었습니다.

def solution(n, control):
    answer = n
    c = { 'w':1, 's':-1, 'd':10, 'a':-10}
    for i in control:
        answer += c[i]
    return answer
```

그리고 매번 조작을 할 때마다 결괏값을 기록한 정수 배열이 numLog입니다. 즉, numLog[i]는 numLog[0]로부터 총 i번의 조작을 가한 결과가 저장되어 있습니다.

주어진 정수 배열 numLog에 대해 조작을 위해 입력받은 문자열을 return 하는 solution 함수를 완성해 주세요.

```python
# 본인 풀이

def solution(numLog):
    answer = ""
    logs = { 1:'w', -1:'s', 10:'d', -10:'a'}
    
    for i in range(len(numLog)):
        if i+1 == len(numLog):
            break
        getter = numLog[i+1] - numLog[i]
        answer += logs[getter]
    
    return answer
```

# 수열과 구간 쿼리 2

- 정수 배열 arr와 2차원 정수 배열 queries이 주어집니다. queries의 원소는 각각 하나의 query를 나타내며, [s, e, k] 꼴입니다.

- 각 query마다 순서대로 s ≤ i ≤ e인 모든 i에 대해 k보다 크면서 가장 작은 arr[i]를 찾습니다.

- 각 쿼리의 순서에 맞게 답을 저장한 배열을 반환하는 solution 함수를 완성해 주세요.
- 단, 특정 쿼리의 답이 존재하지 않으면 -1을 저장합니다.

```python
# 제 풀이입니다.

def solution(arr, queries):
    answer = []
    
    for i, query in enumerate(queries):
        candidate = []
        for num in arr[query[0]:query[1]+1]:
            if num > query[2]:
                candidate.append(num)
        if not candidate:
            answer.append(-1)
        else:
            answer.append(min(candidate))
    
    return answer
```

```python
a = solution([0, 1, 2, 4, 3], [[0, 4, 2],[0, 3, 2],[0, 2, 2]])
print(a)
```

```python
# 다른 사람이 푼 풀이입니다.
# 논리는 같은데, 리스트 컴프리헨션과 기법을 통해 더 간결하게 표현했습니다.

def solution(arr, queries):
    answer = []
    for s, e, k in queries:
        l = [i for i in arr[s:e+1] if i > k]
        answer.append(-1 if len(l) == 0 else min(l))
    return answer
```

- 정수 배열 arr와 2차원 정수 배열 queries이 주어집니다. queries의 원소는 각각 하나의 query를 나타내며, [s, e, k] 꼴입니다.

- 각 query마다 순서대로 s ≤ i ≤ e인 모든 i에 대해 i가 k의 배수이면 arr[i]에 1을 더합니다.

- 위 규칙에 따라 queries를 처리한 이후의 arr를 return 하는 solution 함수를 완성해 주세요.

```python
def solution(arr, queries):
    for query in queries:
        s, e, k = query
        for pt in range(s, e+1):
            if pt==0:
                arr[0] += 1
            elif pt % k == 0:
                arr[pt] += 1
    return arr
```

# 배열 만들기 4

정수 배열 arr가 주어집니다. arr를 이용해 새로운 배열 stk를 만드려고 합니다.

변수 i를 만들어 초기값을 0으로 설정한 후 i가 arr의 길이보다 작으면 다음 작업을 반복합니다.

- 만약 stk가 빈 배열이라면 arr[i]를 stk에 추가하고 i에 1을 더합니다.
- stk에 원소가 있고, stk의 마지막 원소가 arr[i]보다 작으면 arr[i]를 stk의 뒤에 추가하고 i에 1을 더합니다.
- stk에 원소가 있는데 stk의 마지막 원소가 arr[i]보다 크거나 같으면 stk의 마지막 원소를 stk에서 제거합니다.

위 작업을 마친 후 만들어진 stk를 return 하는 solution 함수를 완성해 주세요.

```python
# 본인 풀이

def solution(arr):
    stk = []
    i = 0
    while i < len(arr):
        if not stk:
            stk.append(arr[i])
            i += 1
        elif stk[-1] < arr[i]:
            stk.append(arr[i])
            i += 1
        else:
            stk.pop()
    return stk
```

```python
# 더 짧은 풀이가 있습니다.
# 빼는 것만 특정화시키고 그 외의 경우는 다 추가해버립니다.

def solution(arr):
    stk = []
    for i in range(len(arr)):
        while stk and stk[-1] >= arr[i]:
            stk.pop()
        stk.append(arr[i])
    return stk
```

# 주사위 게임 3

1부터 6까지 숫자가 적힌 주사위가 네 개 있습니다. 네 주사위를 굴렸을 때 나온 숫자에 따라 다음과 같은 점수를 얻습니다.

- 네 주사위에서 나온 숫자가 모두 p로 같다면 1111 × p점을 얻습니다.
= 세 주사위에서 나온 숫자가 p로 같고 나머지 다른 주사위에서 나온 숫자가 q(p ≠ q)라면 (10 × p + q)2 점을 얻습니다.
- 주사위가 두 개씩 같은 값이 나오고, 나온 숫자를 각각 p, q(p ≠ q)라고 한다면 (p + q) × |p - q|점을 얻습니다.
- 어느 두 주사위에서 나온 숫자가 p로 같고 나머지 두 주사위에서 나온 숫자가 각각 p와 다른 q, r(q ≠ r)이라면 q × r점을 얻습니다.
- 네 주사위에 적힌 숫자가 모두 다르다면 나온 숫자 중 가장 작은 숫자 만큼의 점수를 얻습니다.

네 주사위를 굴렸을 때 나온 숫자가 정수 매개변수 a, b, c, d로 주어질 때, 얻는 점수를 return 하는 solution 함수를 작성해 주세요.

```python
# 제시한 네 경우를 그냥 모두 조건식을 만들어서 풀었습니다.

def solution(a, b, c, d):
    dices = [a, b, c, d]
    answer = 0
    if len(set(dices)) == 1:
        answer = 1111*a
    elif len(set(dices)) == 2:
        for dice in dices:
            if dices.count(dice) == 3:
                answer = (10*dice + sum(dices)-3*dice)**2
                break
            if dices.count(dice) == 2:
                dices = list(set(dices))
                answer = (dices[0] + dices[1]) * abs(dices[0] - dices[1])
                break
    elif len(set(dices)) == 3:
        for dice in dices:
            if dices.count(dice) == 2:
                dices.remove(dice)
                dices.remove(dice)
                answer = dices[0] * dices[1]
                break
    else:
        answer = min(dices)
    return answer
    
```

# 글자 이어 붙여 문자열 만들기

- 문자열 my_string과 정수 배열 index_list가 매개변수로 주어집니다. 
- my_string의 index_list의 원소들에 해당하는 인덱스의 글자들을 순서대로 이어 붙인 문자열을 return 하는 solution 함수를 작성해 주세요.

제한사항
- 1 ≤ my_string의 길이 ≤ 1,000
- my_string의 원소는 영소문자로 이루어져 있습니다.
- 1 ≤ index_list의 길이 ≤ 1,000
- 0 ≤ index_list의 원소 < my_string의 길이

```python
def solution(my_string, index_list):
    answer = ""
    for i in index_list:
        answer += my_string[i]
    return answer
```

# 문자열 여러 번 뒤집기

- 문자열 my_string과 이차원 정수 배열 queries가 매개변수로 주어집니다. queries의 원소는 [s, e] 형태로, my_string의 인덱스 s부터 인덱스 e까지를 뒤집으라는 의미입니다. my_string에 queries의 명령을 순서대로 처리한 후의 문자열을 return 하는 solution 함수를 작성해 주세요.

제한사항
- my_string은 영소문자로만 이루어져 있습니다.
- 1 ≤ my_string의 길이 ≤ 1,000
- queries의 원소는 [s, e]의 형태로 0 ≤ s ≤ e < my_string의 길이를 만족합니다.
- 1 ≤ queries의 길이 ≤ 1,000

```python
def solution(my_string, queries):
    for query in queries:
        my_string[query[0], query[1]] = my_string[query[0], query[1], -1]
    return my_string
```

```python
# 풀기는 이렇게 풀었습니다.

def solution(my_string, queries):
    my_list = list(my_string)
    for query in queries:
        if not query[0]:
            my_list[:query[1]+1] = my_list[query[1]::-1]
        else:
            my_list[query[0]:query[1]+1] = my_list[query[1]:query[0]-1:-1]
    return "".join(my_list)
```
